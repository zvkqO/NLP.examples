# Torch Tensor
텐서 생성

```python
# 0으로 채운 Tensor
zero_tensor = torch.zeros(3, 3)
print(zero_tensor)

# 1로 채운 Tensor
ones_tensor = torch.ones(3, 3)
print(ones_tensor)

# 랜덤한 값으로 채운 Tensor
random_tensor = torch.rand(3, 3)
print(random_tensor)

# 특정 값을 갖는 Tensor
float_tensor = torch.full((3, 3), -0.7)
print(float_tensor)
```

텐서 연산

```python
x = torch.tensor([1, 2, 3])
y = torch.tensor([4, 5, 6])

# 덧셈
add_result = x + y
print("Addition Result:", add_result)

# 뺄셈
sub_result = x - y
print("Subtraction Result:", sub_result)

# 곱셈 (element-wise)
mul_result = x * y
print("Multiplication Result:", mul_result)

# 나눗셈 (element-wise)
div_result = x / y
print("Division Result:", div_result)
```

```python
a = torch.tensor([[1, 2], [3, 4]])
b = torch.tensor([[5, 6], [7, 8]])

# 행렬곱
matmul_result = torch.matmul(a, b)
print(matmul_result)

# 전치 행렬
transpose_result = a.T
print(transpose_result)
```

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

try:
    dot_product = torch.dot(a, b)
    print(dot_product)
except RuntimeError as e:
    print("Error:", e)
```

브로드캐스팅

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])

print(A + 2)
```

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])

print(np.matmul(A, B))
```

```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8, 9], [10, 11, 12]])

try:
    matmul_result = np.matmul(A, B)
    print(matmul_result)
except ValueError as e:
    print("Error:", e)
```

```python
a = torch.tensor([[0, 0], [0, 0]]) # (2, 2)
b = torch.tensor([1, 2]) # (2,) -> (2,2)로 늘려서 연산
c = torch.tensor([[1], [2]]) # (1, 2) -> (2,2)로 늘려서 연산

print(a + b)
print(a + c)
```

```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]]) # (2, 3)
b = torch.tensor([10, 20, 30]) # (3,) -> (2, 3)으로 확장

print(a * b)
```

```python
import torch

A = torch.ones(2, 2, 2) # (2, 2, 2)
B = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32) # (2, 2) -> (2, 2, 2)로 확장됨

print(A)
print(torch.matmul(A, B))
```

```python
a = torch.tensor([[1, 2], [3, 4]]) # (2, 2)
b = torch.tensor([1, 2, 3]) # (3,)

try:
    print(a + b)
except RuntimeError as e:
    print("Error:", e)
```

```python
A = torch.ones(2, 2)  # Shape: (2, 2)
B = torch.tensor([1, 2], dtype=torch.float32)  # Shape: (2,)

print(torch.matmul(A, B))
```
